create a reusable Python function that safely parses incoming contract execution webhooks (simulating a Docusign or enterprise platform event)

```python
def parse_webhook_event(payload):
    envelope_id = payload.get("envelope_id", "UNKNOWN_ENV")
    raw_status = payload.get("status", "UNKNOWN")

    clean_status = raw_status.lower()

    is_completed = (clean_status == "completed")

    return {
        "id": envelope_id,
        "status": clean_status,
        "is_completed": is_completed
    }

sample_webhooks = [
    {"envelope_id": "env-101", "status": "COMPLETED"},
    {"envelope_id": "env-102", "status": "DECLINED"},
    {"envelope_id": "env-103"},  # Missing status key!
    {"status": "SENT"}           # Missing envelope_id key!
]

for webhook in sample_webhooks:
    parsed_result = parse_webhook_event(webhook)
    print(parsed_result)
    
```
