```python
scores = [0.95, 0.88, 0.65, 0.92, 0.40]

passing_scores = []
rejected_count = 0

for score in scores:
    if score >= 0.80:
        passing_scores.append(score)
    else:
        rejected_count = rejected_count + 1

print("Passing scores:", passing_scores)
print("Rejected count:", rejected_count)
```
