basic data type

```python
scores = [0.95, 0.88, 0.65, 0.92, 0.40]

passing_scores = []
rejected_count = 0

for score in scores:
    if score >= 0.80:
        passing_scores.append(score)
    else:
        rejected_count = rejected_count + 1

print("Passing scores:", passing_scores)
print("Rejected count:", rejected_count)
```

Webhook Event Filter

```python
webhook_payloads = [
    {"envelope_id": "env-001", "status": "COMPLETED", "signer": "Alice"},
    {"envelope_id": "env-002", "status": "declined", "signer": "Bob"},
    {"envelope_id": "env-003", "status": "SENT", "signer": "Charlie"},
    {"envelope_id": "env-004", "status": "COMPLETED", "signer": "David"}
]
for webhook in webhook_payloads:
    if "status" in webhook:
        if webhook["status"].lower() == "completed":
            print(webhook["envelope_id"])
```
